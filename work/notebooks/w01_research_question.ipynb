## 2. The question: decision, action, cost of a wrong call

**Research Question:** Which content pages should be reviewed first for refresh based on observable search and engagement signals?

- **Decision:** Decide which pages should be reviewed first.
- **Who acts?** Content editors or the SEO team.
- **Unit of analysis:** One content page.
- **Output:** A ranked refresh priority score.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

df = pd.read_csv("/content/content_refresh_anonymized.csv")

print("Dataset shape:", df.shape)
print("\nTrend Direction:")
print(df["trend_direction"].value_counts())

print("\nAverage impressions:")
print(df["impressions_90d"].mean())

print("\nAverage sessions:")
print(df["sessions_90d"].mean())
```

The starter dataset contains **30,000 content pages** with **44 columns**, providing enough examples for exploratory analysis.

The distribution of `trend_direction` shows that **16,262 pages are currently marked as "down"**, indicating that many pages may require review, while other pages are classified as stable, up, new, or flat.

The average **90-day impressions are approximately 5,200**, and the average **90-day sessions are about 37**, showing that many pages have measurable search visibility and user engagement.
