# ML-02 — Research Question and Provisional Lane


## 1. My Lane and Why

**Freestyle Capstone Title:**
Do AI-Assistant Referral Signals Improve Content Decay Prediction? A Multi-Architecture Ablation Study

**Project Rationale:**
Traditional SEO audits rely heavily on lagging Google Search Console indicators to spot declining performance. With modern search behavior increasingly influenced by conversational AI discovery (ChatGPT, Claude, Gemini, Copilot, Perplexity), a common assumption is that AI-assistant traffic serves as an immediate leading indicator for content health.

This project sets out to rigorously test that hypothesis through a multi-channel early warning system and ablation study. Rather than assuming AI traffic shifts act as active predictive drivers, this pipeline evaluates whether incorporating AI-assistant referral sessions *actually* improves the predictive accuracy of content decay models beyond what traditional search and engagement metrics achieve. By putting the AI referral hypothesis through rigorous multi-model validation (comparing architectures like HistGradientBoosting and Random Forest across cross-validation), this study provides an honest, data-driven assessment of multi-channel signals for editorial decision support.

## 2. The Question: Decision, Action, Cost of a Wrong Call

* **Research Question:** Does incorporating AI-assistant referral signals (ChatGPT, Claude, Gemini, Copilot, Perplexity) improve the predictive accuracy of a multi-channel content decay early warning system, beyond what traditional search and engagement metrics alone provide?
* **Decision:** Determine which content assets require immediate manual review and intervention versus those maintaining stable performance.
* **Who Acts:** Content editors, SEO managers, and growth strategists.
* **Unit of Analysis:** A single unique content page (`content_id`).
* **Output:** A calibrated risk probability score and a structured, tier-based priority action queue.
* **Actionable Playbook:**
  - **P1 (Full Overhaul):** Complete content restructuring and semantic rewrite for high-decay risk pages.
  - **P2 (CTR & Title Refresh):** Meta tag and title optimization for pages experiencing visibility drift.
  - **P3 / P4 (Monitor / Internal Links):** Routine monitoring or internal linking adjustments.
* **Cost of a Wrong Recommendation:**
  - *False Positive:* Misallocation of editorial resources toward updating a page that is structurally healthy.
  - *False Negative:* Failure to detect a decaying asset, resulting in an unmitigated loss of both traditional search visibility and emerging AI-assistant referral traffic.
* **Why Machine Learning:** Machine learning can model non-linear interactions among multi-channel signals that would be impractical to specify manually as fixed rules at this scale.

## 4. Careful Words: What I Can and Can't Claim

* **What this work CAN claim:**

    This system delivers **observational, directional decision-support**. It captures historical correlations between multi-channel traffic signals (including AI-assistant referrals) and subsequent content performance degradation, ranking pages to optimize human editorial workflows.
* **What this work CANNOT claim:**

    This modeling framework does **not** establish causal proof (i.e., we cannot assert that shifts in AI referrals directly *cause* search traffic drops, only that they serve as predictive indicators).
